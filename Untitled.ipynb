```python
import matplotlib.pyplot as plt
import seaborn as sns
import numpy as np

features_to_compare = ["dur", "sbytes", "dbytes", "spkts", "dpkts"]

# Prepare a copy for visualization
plot_df = train_df[features_to_compare + ["label"]].copy()

# Convert labels to readable names
plot_df["Traffic Type"] = plot_df["label"].map({
    0: "Normal",
    1: "Attack"
})

# Log transformation for visualization
for feature in features_to_compare:
    plot_df[feature] = np.log1p(plot_df[feature])

# Convert to long format
plot_data = plot_df.melt(
    id_vars="Traffic Type",
    value_vars=features_to_compare,
    var_name="Network Feature",
    value_name="Log-Transformed Value"
)

# Plot
plt.figure(figsize=(14, 7))

sns.boxplot(
    data=plot_data,
    x="Network Feature",
    y="Log-Transformed Value",
    hue="Traffic Type",
    palette={"Normal": "#36CFC9", "Attack": "#FF6B6B"},
    showfliers=False
)

plt.title(
    "Network Feature Comparison: Normal vs Attack Traffic",
    fontsize=18,
    fontweight="bold",
    pad=20
)

plt.xlabel("Network Feature", fontsize=12)
plt.ylabel("Log(1 + Feature Value)", fontsize=12)

plt.legend(title="Traffic Type")
plt.tight_layout()

plt.savefig(
    "EDA_Normal_vs_Attack_Features.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import numpy as np
import pandas as pd

radar_features = ["dur", "sbytes", "dbytes", "spkts", "dpkts"]

# Exclude Normal traffic
attack_data = train_df[
    train_df["attack_cat"].str.lower() != "normal"
].copy()

# Calculate mean values by attack category
radar_data = attack_data.groupby("attack_cat")[radar_features].mean()

# Log transform to reduce skew
radar_data = np.log1p(radar_data)

# Normalize each feature between 0 and 1
radar_data = (
    (radar_data - radar_data.min()) /
    (radar_data.max() - radar_data.min())
)

radar_data
```

```python
import matplotlib.pyplot as plt

categories = radar_features
angles = np.linspace(0, 2 * np.pi, len(categories), endpoint=False).tolist()

angles += angles[:1]

fig, ax = plt.subplots(figsize=(10, 10), subplot_kw={"polar": True})

colors = plt.cm.tab10(np.linspace(0, 1, len(radar_data)))

for i, (attack, values) in enumerate(radar_data.iterrows()):
    values = values.tolist()
    values += values[:1]

    ax.plot(
        angles,
        values,
        linewidth=2,
        label=attack,
        color=colors[i]
    )

    ax.fill(
        angles,
        values,
        alpha=0.08,
        color=colors[i]
    )

ax.set_xticks(angles[:-1])
ax.set_xticklabels(categories, fontsize=11)

ax.set_title(
    "Network Behavior Profile Across Attack Categories",
    fontsize=17,
    fontweight="bold",
    pad=25
)

ax.legend(
    loc="upper right",
    bbox_to_anchor=(1.3, 1.15),
    fontsize=9
)

plt.tight_layout()

plt.savefig(
    "EDA_Attack_Category_Radar.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

protocol_label = pd.crosstab(
    train_df["proto"],
    train_df["label"],
    normalize="index"
) * 100

# Select the 15 most common protocols
top_protocols = train_df["proto"].value_counts().head(15).index
protocol_label = protocol_label.loc[top_protocols]

protocol_label = protocol_label.rename(
    columns={0: "Normal Traffic", 1: "Attack Traffic"}
)

protocol_label = protocol_label.sort_values(
    "Attack Traffic",
    ascending=True
)

plt.figure(figsize=(12, 8))

protocol_label.plot(
    kind="barh",
    stacked=True,
    figsize=(12, 8),
    color=["#36CFC9", "#FF6B6B"],
    edgecolor="white",
    linewidth=0.5
)

plt.title(
    "Protocol-wise Normal vs Attack Traffic",
    fontsize=17,
    fontweight="bold",
    pad=15
)

plt.xlabel("Traffic Composition (%)")
plt.ylabel("Network Protocol")
plt.legend(title="Traffic Type", loc="lower right")
plt.xlim(0, 100)

plt.tight_layout()
plt.savefig("EDA_Protocol_Traffic_Composition.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd

# Exclude normal traffic
attack_df = train_df[
    train_df["attack_cat"].str.lower() != "normal"
].copy()

# Select top 10 protocols and attack categories
top_protocols = attack_df["proto"].value_counts().head(10).index
top_attacks = attack_df["attack_cat"].value_counts().head(10).index

filtered_df = attack_df[
    attack_df["proto"].isin(top_protocols) &
    attack_df["attack_cat"].isin(top_attacks)
]

heatmap_data = pd.crosstab(
    filtered_df["attack_cat"],
    filtered_df["proto"],
    normalize="index"
) * 100

plt.figure(figsize=(14, 8))

sns.heatmap(
    heatmap_data,
    cmap="YlOrRd",
    annot=True,
    fmt=".1f",
    linewidths=0.5,
    cbar_kws={"label": "Percentage (%)"}
)

plt.title(
    "Attack Category Distribution Across Network Protocols",
    fontsize=17,
    fontweight="bold",
    pad=18
)

plt.xlabel("Network Protocol")
plt.ylabel("Attack Category")

plt.tight_layout()
plt.savefig(
    "EDA_Protocol_Attack_Category_Heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
import pandas as pd
import numpy as np
import plotly.express as px

from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA

# Select numerical features
numeric_data = train_df.select_dtypes(include=["int64", "float64"]).copy()

# Remove ID and target-related columns
numeric_data = numeric_data.drop(
    columns=["id", "label"],
    errors="ignore"
)

# Handle infinite values
numeric_data = numeric_data.replace(
    [np.inf, -np.inf],
    np.nan
)

numeric_data = numeric_data.fillna(numeric_data.median())

# Sample records for faster visualization
sample_size = min(15000, len(train_df))

sample_indices = train_df.sample(
    n=sample_size,
    random_state=42
).index

X_sample = numeric_data.loc[sample_indices]
y_sample = train_df.loc[sample_indices, "label"]

# Standardize features
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X_sample)

# Apply PCA
pca = PCA(n_components=2)
pca_result = pca.fit_transform(X_scaled)

pca_df = pd.DataFrame({
    "PCA Component 1": pca_result[:, 0],
    "PCA Component 2": pca_result[:, 1],
    "Traffic Type": y_sample.map({
        0: "Normal Traffic",
        1: "Attack Traffic"
    }).values
})

# Interactive scatter visualization
fig = px.scatter(
    pca_df,
    x="PCA Component 1",
    y="PCA Component 2",
    color="Traffic Type",
    color_discrete_map={
        "Normal Traffic": "#00F5D4",
        "Attack Traffic": "#FF4D6D"
    },
    opacity=0.65,
    title="UNSW-NB15 | PCA Visualization of Network Traffic",
    template="plotly_dark",
    render_mode="webgl"
)

fig.update_traces(
    marker=dict(size=5),
    hovertemplate=(
        "<b>%{fullData.name}</b><br>"
        "PCA 1: %{x:.3f}<br>"
        "PCA 2: %{y:.3f}<extra></extra>"
    )
)

fig.update_layout(
    title=dict(
        text="UNSW-NB15 | Normal vs Attack Traffic in PCA Space",
        x=0.5,
        font=dict(size=22)
    ),
    font=dict(family="Arial", size=13),
    legend_title="Traffic Class",
    xaxis_title="Principal Component 1",
    yaxis_title="Principal Component 2",
    height=750
)

fig.show()

fig.write_html("EDA_PCA_Traffic_Visualization.html")
```
